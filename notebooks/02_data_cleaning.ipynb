# Data Cleaning – Superstore Sales Dataset

## Purpose
This notebook applies the necessary transformations identified
during the data profiling phase to prepare the dataset for analysis.

1️⃣ Import libraries

```python
import pandas as pd
```

2️⃣ Load dataset

```python
df_raw = pd.read_csv("../data/raw/superstore_raw.csv")
df = df_raw.copy()
```

3️⃣ Dataset Overview Before Cleaning

```python
df.info()
```

5️⃣ Convert Date Columns

Order Date and Ship Date were stored as object type.

They will be converted to datetime format to enable time-based analysis.

```python
df["Order Date"] = pd.to_datetime(df["Order Date"], format="%d/%m/%Y")
df["Ship Date"] = pd.to_datetime(df["Ship Date"], format="%d/%m/%Y")
```

6️⃣ Handle Missing Postal Codes

Only 11 records contain missing values in the Postal Code column.
Given the low percentage, records will be retained and missing
values will be labeled as "Unknown" to preserve dataset integrity.

```python
df["Postal Code"] = df["Postal Code"].fillna("Unknown")
```

7️⃣ Standardize Data Types


Postal Code will be converted to string format to prevent
misinterpretation as a numerical variable.

```python
df["Postal Code"] = df["Postal Code"].astype(str)
df["Postal Code"] = df["Postal Code"].str.replace(".0", "", regex=False)
```

8️⃣ Final Dataset Check

```python
df.info()
df.isnull().sum()
```

9️⃣ Cleaning Summary

The dataset was cleaned according to the findings from the
data profiling phase.

Actions performed:

- Converted date columns to datetime format
- Handled missing Postal Codes
- Standardized data types

The dataset is now structurally consistent and ready for
exploratory data analysis.

🔟 Basic Feature Engineering - Shipping Lead Time

A new feature will be created to measure the number of days
between Order Date and Ship Date.

This metric is relevant for operational efficiency analysis
and service-level evaluation.

```python
df["Lead Time (Days)"] = (df["Ship Date"] - df["Order Date"]).dt.days
df["Lead Time (Days)"].describe()
```

```python
# Check for negative lead times
df[df["Lead Time (Days)"] < 0]
```

1️⃣1️⃣ Basic Featute Engineering - Year / Month

Year and Month features are created to facilitate
trend analysis and time-series aggregation.

```python
df["Order Year"] = df["Order Date"].dt.year
df["Order Month"] = df["Order Date"].dt.month
df["Order Month Name"] = df["Order Date"].dt.month_name()

# For Power BI, we can also create a "Year-Month" column
df["Order Year-Month"] = df["Order Date"].dt.to_period("M").astype(str)
```

```python
df.head()
```

1️⃣2️⃣ Order-Level Consistency Check

Since an Order ID can contain multiple products,
key attributes such as Order Date, Customer ID,
and Ship Mode should remain consistent within each order.

```python
order_consistency = df.groupby("Order ID").agg(
    {
        "Order Date": "nunique",
        "Customer ID": "nunique",
        "Ship Mode": "nunique",
    }
)

order_consistency[
    (order_consistency["Order Date"] > 1) | (order_consistency["Customer ID"] > 1) | (order_consistency["Ship Mode"] > 1)
]
```

1️⃣3️⃣ Enhanced Data Preparation Summary

In addition to structural cleaning, feature engineering
was applied to enhance analytical capability.

New features created:

- Shipping Lead Time
- Order Year / Month
- Year-Month aggregation key
- Order-level consistency validation

The dataset is now analytics-ready and structured
for both business intelligence and operational insights.
