# Exploratory Data Analysis – Superstore Sales Dataset

## Purpose
This notebook explores sales performance,
customer behavior, and operational efficiency
to generate business-driven insights.

1️⃣ Import libraries

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

2️⃣ Load clean dataset

```python
df = pd.read_csv("../data/processed/superstore_cleaned.csv")
df["Order Date"] = pd.to_datetime(df["Order Date"])
```

3️⃣ Sección 1 – Executive Overview


This section evaluates total revenue,
distribution patterns, and overall sales concentration.

```python
df["Sales"].describe().round(2)
```

```python
plt.figure()
sns.histplot(df["Sales"].to_frame(), bins=50)
plt.title("Sales Distribution")
plt.show()
```

4️⃣ Sección 2 – Time Analysis

Understanding seasonality and growth patterns.

```python
# Ventas por Year-Month
monthly_sales = df.groupby("Order Year-Month")["Sales"].sum()
monthly_sales.plot()
plt.title("Monthly Sales Trend")
plt.show()
```

5️⃣ Sección 3 – Segment & Region Performance

Identifying high-performing and underperforming areas.

```python
# Ventas por segmento
df.groupby("Segment")["Sales"].sum().sort_values(ascending=False).plot(kind="bar")
plt.title("Sales by Segment")
```

```python
# Ventas por región
df.groupby("Region")["Sales"].sum().sort_values(ascending=False).plot(kind="bar")
plt.title("Sales by Region")
```

6️⃣ Sección 4 – Operational Efficiency

Evaluating operational performance and delivery efficiency.

```python
# Lead Time promedio
df["Lead Time (Days)"].describe().round(2)
```

```python
# Lead Time por Ship Mode
df.groupby("Ship Mode")["Lead Time (Days)"].mean().round(2)
```

7️⃣ Sección 5 – Category & Product Insights

```python
df.groupby("Category")["Sales"].sum().round(2)
```

```python
df.groupby("Sub-Category")["Sales"].sum().sort_values(ascending=False).round(2)
```

8️⃣ Sección Final – Business Insights Summary

1. Sales Distribution Insight

Sales are highly right-skewed, with the majority of transactions
concentrated at low monetary values and a small number of
high-value outliers.

This suggests that revenue is likely driven by a combination
of frequent low-value purchases and occasional large transactions.

2. Monthly Sales Trend Insight

Sales show an overall upward trend over time, indicating
business growth. However, significant month-to-month
volatility is present, with recurring sales peaks that may
suggest seasonality or promotional cycles.

The highest revenue levels are observed in the final year,
indicating potential market expansion or improved performance.

3. 

    - Segment Performance Insight

The Consumer segment generates the majority of total revenue,
significantly outperforming Corporate and Home Office segments.
This indicates that the business is primarily driven by
individual customers rather than institutional buyers.

    - Regional Performance Insight

The West region leads total sales, closely followed by the East.
In contrast, the South region contributes the lowest revenue,
suggesting potential growth opportunities or market limitations
in that area.

4. Shipping Lead Time Insight

Lead time increases progressively from Same Day to Standard Class,
indicating consistent operational performance aligned with
shipping priority levels.

Standard Class averages approximately 5 days, while expedited
options significantly reduce delivery time. This trade-off
between cost and speed may influence customer satisfaction
and purchasing behavior.

5. 

    - Category Performance Insight

Technology generates the highest total revenue, followed by
Furniture and Office Supplies. Sales are relatively well
distributed across categories, reducing dependence on a
single product line.

    - Sub-category Performance Insight

Revenue is concentrated in a small number of sub-categories,
particularly Phones and Chairs, which represent the primary
drivers of sales. In contrast, several sub-categories contribute
marginal revenue, suggesting potential product optimization
or inventory rationalization opportunities.


#### These insights can inform pricing strategy, inventory planning, and logistics optimization.
